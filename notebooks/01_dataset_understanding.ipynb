# Understanding the FinBridge files
Each training row describes one historical loan application and its outcome. The test file withholds that outcome. Run from the project root or the notebooks folder.

```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
from IPython.display import Markdown, display

ROOT = Path.cwd()
if not (ROOT / "train.csv").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
train = pd.read_csv(ROOT / "train.csv")
test = pd.read_csv(ROOT / "test.csv")
```

## Inspect both ends of each file
The first and last five rows show the raw formatting and help confirm that the target appears only in training.

```python
for name, frame in [("Training", train), ("Test", test)]:
    display(Markdown(f"### {name} · {frame.shape[0]:,} rows × {frame.shape[1]} columns"))
    display(frame.head(), frame.tail())
```

## Schema and column roles
Pandas reads the two percent fields as text. Their financial values are numerical. Identifiers match records and the target describes an outcome.

```python
ids = ["application_id", "applicant_ref_id"]
target = "default_flag"
inputs = train.drop(columns=target)
numeric = [col for col in inputs.select_dtypes("number") if col not in ids]
numeric += ["interest_rate_pct", "revolving_utilization_pct"]
categorical = [col for col in inputs if col not in ids + numeric]
schema = pd.DataFrame({"train_dtype": train.dtypes, "test_dtype": test.dtypes})
schema["role"] = ["identifier" if c in ids else "numerical" if c in numeric
                  else "categorical" if c in categorical else "target" for c in schema.index]
display(schema)
print("Ordered inputs match:", inputs.columns.equals(test.columns))
print("Training only:", inputs.columns.difference(test.columns).tolist())
print("Test only:", test.columns.difference(inputs.columns).tolist())
```

## Check record identity
Count repeated complete rows and repeated application keys separately. Also check for shared keys between files.

```python
for name, frame in [("Train", train), ("Test", test)]:
    print(name, "duplicate rows:", frame.duplicated().sum(),
          "| duplicate application IDs:", frame.application_id.duplicated().sum())
print("Shared application IDs:", len(set(train.application_id) & set(test.application_id)))
```

## Repayment outcomes
A value of 1 means default or charge-off and 0 means full repayment.

```python
counts = train[target].value_counts(dropna=False).sort_index()
display(pd.DataFrame({"applications": counts, "percent": 100 * counts / len(train)}).round(2))
```

The files contain 31,016 training applications and 7,754 test applications, with the same 22 input columns. Of the training loans, 4,507 defaulted (14.53%) and 26,509 were repaid.
